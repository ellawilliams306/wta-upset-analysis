# Feature engineering for WTA match data

This is the feature engineering part of the project. We describe the role our current columns will have in our model either as model features, the target, or reporting only data.

| Column                              | Role                 | Reason                    |
| ----------------------------------- | -------------------- | ------------------------- |
| `is_upset`                          | Target               | Match outcome to predict  |
| `rank_gap` / `log_rank_gap`         | Numerical feature    | Known before the match    |
| `surface`, `round`, `tourney_level` | Categorical features | Known before the match    |
| Names and ranks                     | Reporting only       | Never passed to the model |


We choose to not to keep any of the 2020 matches when creating the model data, since the 2020 WTA season was significantly disrupted due to the Covid-19 pandemic, and it was not representative of a typical season.

We organise this notebook as follows.


1. Load cleaned match data
2. Define the prediction target
3. Choose pre-match features
4. Create and check those features
5. Explore numerical relationships
6. Save the model-ready dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
matches = pd.read_csv("../data/processed/matches_clean.csv")
```

```python
model_columns = [
    "tourney_date",
    "tourney_name",
    "surface",
    "round",
    "tourney_level",
    "winner_name",
    "winner_rank",
    "loser_name",
    "loser_rank",
    "rank_gap",
    "is_upset",
    "year"
]

model_data = matches[model_columns].dropna(
    subset=[
        "tourney_date",
        "surface",
        "round",
        "tourney_level",
        "rank_gap",
        "is_upset",
        "year"
    ]
).copy()

model_data["is_upset"] = model_data["is_upset"].astype(int)
```

```python
model_data["log_rank_gap"] = np.log1p(
    model_data["rank_gap"]
)

model_data.head()
```

## Modelling scope

The 2020 season is excluded from predictive modelling because its disrupted
schedule and reduced number of matches make it unrepresentative of a typical
WTA season. It remains in the exploratory dataset.

```python
model_data = model_data.loc[
    model_data["year"] != 2020
].copy()
```

Finally, we again save our data ready for the baseline model.

```python
model_data.to_csv("../data/processed/model_data_baseline.csv",index=False)
```
